## Experimento aleatorio: lanzamiento de dos dados
El espacio muestral está formado por las 36 parejas ordenadas posibles.


```python
from itertools import product
from fractions import Fraction

Omega = list(product(range(1, 7), repeat=2))
print(Omega)
print("Número de resultados:", len(Omega))
```

## Eventos aleatorios
S(n) contiene los resultados donde la suma de los dados es n.


```python
def S(n):
    return [(i, j) for i, j in Omega if i + j == n]

for n in range(2, 13):
    print(f"S({n}) =", S(n))
```

## Probabilidad de obtener suma 9


```python
favorables = S(9)
probabilidad = Fraction(len(favorables), len(Omega))
print("Resultados favorables:", favorables)
print("P(suma = 9) =", probabilidad)
print("Valor decimal =", float(probabilidad))
```

## Conclusión
En dos dados hay 36 resultados posibles. Para una suma igual a 9 existen 4 casos favorables, por lo que la probabilidad es 4/36 = 1/9.
