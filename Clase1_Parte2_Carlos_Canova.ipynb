## Lanzamiento de tres monedas
A representa águila y S representa sol.


```python
from itertools import product
from fractions import Fraction
import random

Omega = set(product(["A", "S"], repeat=3))
print(Omega)
print("Número de resultados:", len(Omega))
```

## Eventos por posición


```python
A_1 = {om for om in Omega if om[0] == "A"}
A_2 = {om for om in Omega if om[1] == "A"}
A_3 = {om for om in Omega if om[2] == "A"}
print("A1 =", A_1)
print("A2 =", A_2)
print("A3 =", A_3)
```

## Probabilidad de tres águilas


```python
def P(evento):
    return Fraction(len(evento), len(Omega))

E = {om for om in Omega if om.count("A") == 3}
print("E =", E)
print("P(E) =", P(E))
```

## Águila en los dos primeros lanzamientos


```python
F = A_1.intersection(A_2)
print("F =", F)
print("P(F) =", P(F))
```

## Simulación de una moneda no equilibrada


```python
def volado(p):
    return "sol" if random.random() < p else "aguila"

random.seed(24)
resultados = [volado(0.4) for _ in range(10)]
print(resultados)
```

```python
random.seed(24)
resultados_07 = [volado(0.7) for _ in range(10)]
print(resultados_07)
```

## Conclusión
Python permite representar eventos mediante conjuntos y simular experimentos aleatorios con probabilidades diferentes.
